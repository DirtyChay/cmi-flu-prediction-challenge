## 1. Imports & Data Loading <a id='1'></a>

```python
from itertools import combinations

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# sns.set_theme(style='whitegrid', palette='muted')
# plt.rcParams['figure.dpi'] = 120
```

```python
DATA_PATH = 'data/PART2-26-01-26_reorg/PART2-26-01-26_reorg'

train_array = pd.read_csv(DATA_PATH + '/train_array.tsv', sep='\t')
train_flow = pd.read_csv(DATA_PATH + '/train_flow.tsv', sep='\t', low_memory=False)
train_participants = pd.read_csv(DATA_PATH + '/train_participants.tsv', sep='\t')

tables = {
    'train_array': train_array,
    'train_flow': train_flow,
    'train_participants': train_participants,
}
```

## 2. Shape & Missing Values <a id='2'></a>
For each table we print the row/column counts, then list every column that has at least one missing value, along with the count and percentage.

```python
for name, df in tables.items():
    print(f'===============================')
    print(f'{name}')
    print(f'{df.shape[0]:,} rows, {df.shape[1]} columns')
    print(f'===============================')

    missing_counts = df.isna().sum()
    missing_only = missing_counts[missing_counts > 0].sort_values(ascending=False)
    if len(missing_only) > 0:
        print('Columns with missing values:')
        for col, count in missing_only.items():
            pct = (count / len(df)) * 100
            print(f'  - {col}: {count:,} missing ({pct:.2f}%)')
    else:
        print('  (No missing values found)')
    print('\n')
```

## 6. Categorical Column Summaries <a id='6'></a>
For string/object columns: number of unique values and value counts.
Columns with more than MAX_UNIQUE_TO_SHOW unique values only show the top MAX_MOST_COMMON_TO_SHOW most common.

```python
MAX_UNIQUE_TO_SHOW = 15
MAX_MOST_COMMON_TO_SHOW = 5

for name, df in tables.items():
    cat_cols = df.select_dtypes(include='object').columns
    if len(cat_cols) == 0:
        continue

    print(f"\n{'=' * 50}")
    print(f'TABLE: {name} — Categorical Columns')
    print(f"{'=' * 50}")
    for col in cat_cols:
        n_unique = df[col].nunique()
        print(f'\n  [{col}] — {n_unique} unique values')
        if n_unique <= MAX_UNIQUE_TO_SHOW:
            print(df[col].value_counts(dropna=False).to_string())
        else:
            print(f'  (too many to list — top {MAX_MOST_COMMON_TO_SHOW} shown)')
            print(df[col].value_counts(dropna=False).head(MAX_MOST_COMMON_TO_SHOW).to_string())
```

## 7. Data Quality Issues & Fixes <a id='7'></a>

Issues identified during EDA:

| # | Table | Column | Issue | Fix |
|---|---|---|---|---|
| 1 | `train_participants` | `biological_sex` | Mixed case: `'female'` and `'Female'` treated as different | Lowercase all values |
| 2 | `train_participants` | `race` | `'race: unknown'` / `'Unknown'` duplicates; same for `'race: other'` | Map to canonical names |
| 3 | `train_flow` | `value` | Contains 1 negative value (min = −1), biologically impossible for cell counts | Clip to 0 |
| 4 | `train_flow` | `population_defnition` | Column name is misspelled (`defnition` → should be `definition`) | Noted; left as-is to match raw data |
| 5 | `train_flow` | `value` | Mixed units (`cells/uL`, `percentage`, `percentile`, etc.) — not directly comparable | Document; always filter by `unit` |
| 6 | `train_array` | `platform_id` | 4 different microarray platforms — batch effects likely | Document; ComBat recommended |


```python
# Fix 1: Normalize biological_sex capitalization
train_participants['biological_sex'] = train_participants['biological_sex'].str.lower()
print('biological_sex after fix:')
print(train_participants['biological_sex'].value_counts())
```

```python
# Fix 2: Consolidate redundant race categories
race_map = {
    'race: unknown': 'Unknown',
    'race: other': 'Other'
}
train_participants['race'] = train_participants['race'].replace(race_map)
print('race after fix:')
print(train_participants['race'].value_counts(dropna=False))
```

```python
# Fix 3: Drop rows with negative values in train_flow
# Negative cell counts/percentages are physically meaningless.
# Only 1 row affected (rounding artifact from a computed population subtraction).
# Dropping is preferred over clipping to 0 since it avoids introducing a fabricated value.
neg_mask = train_flow['value'] < 0
print(f'Rows with negative values in train_flow: {neg_mask.sum()}')
if neg_mask.any():
    display(train_flow.loc[neg_mask, ['participant_id', 'timepoint', 'name', 'population_defnition', 'unit', 'value']])

train_flow = train_flow[~neg_mask].reset_index(drop=True)
print(f'Rows after dropping negatives: {len(train_flow):,}')
```

## 8. Timepoint Coverage <a id='8'></a>

Understanding which timepoints are present and how many measurements exist at each one is crucial for longitudinal modeling. Sparse timepoints can cause missing data problems downstream.

Expected timepoints based on the study design: **-14, 0, 1, 3, 7, 28, 90, ~365** days relative to vaccination.

```python
for name, df in tables.items():
    if 'timepoint' not in df.columns:
        continue
    tp_counts = df['timepoint'].value_counts().sort_index()
    print(f'{name} — timepoints present ({df["timepoint"].nunique()} unique):')
    print(tp_counts.to_string())
    print()
```

```python
# Visualize timepoint distributions
tp_tables = {name: df for name, df in tables.items() if 'timepoint' in df.columns}

fig, axes = plt.subplots(1, len(tp_tables), figsize=(14, 4))

for ax, (name, df) in zip(axes, tp_tables.items()):
    tp_counts = df['timepoint'].value_counts().sort_index()
    ax.bar(tp_counts.index.astype(str), tp_counts.values, color='steelblue', edgecolor='white')
    ax.set_title(name, fontsize=10)
    ax.set_xlabel('Timepoint (days post-vaccination)')
    ax.set_ylabel('Number of measurements')
    ax.tick_params(axis='x', rotation=45)

plt.suptitle('Measurement Counts per Timepoint', fontsize=12, y=1.02)
plt.tight_layout()
plt.show()
```

## 9. Participant Overlap Across Tables <a id='9'></a>

Not every participant will have data in all three tables. Knowing the overlap is essential before merging — participants missing from one table will become NaN rows in a join, which can silently break models.

```python
ids = {name: set(df['participant_id']) for name, df in tables.items()}

print('Unique participants per table:')
for name, id_set in ids.items():
    print(f'  {name}: {len(id_set):,}')

print('\nPairwise overlaps:')
for (n1, s1), (n2, s2) in combinations(ids.items(), 2):
    overlap = len(s1 & s2)
    print(f'  {n1} ∩ {n2}: {overlap:,} shared participants')

all_three = ids['train_array'] & ids['train_flow'] & ids['train_participants']
print(f'\nParticipants present in ALL 3 tables: {len(all_three):,}')
```

```python
# Visualize overlap as a Venn-style bar chart
# (venn3 requires matplotlib-venn; this bar chart works without extra installs)
overlap_data = {
    'array only': len(ids['train_array'] - ids['train_flow'] - ids['train_participants']),
    'flow only': len(ids['train_flow'] - ids['train_array'] - ids['train_participants']),
    'participants only': len(ids['train_participants'] - ids['train_array'] - ids['train_flow']),
    'array ∩ flow': len((ids['train_array'] & ids['train_flow']) - ids['train_participants']),
    'array ∩ participants': len((ids['train_array'] & ids['train_participants']) - ids['train_flow']),
    'flow ∩ participants': len((ids['train_flow'] & ids['train_participants']) - ids['train_array']),
    'all 3': len(all_three),
}

fig, ax = plt.subplots(figsize=(10, 4))
ax.barh(list(overlap_data.keys()), list(overlap_data.values()), color='coral', edgecolor='white')
ax.set_xlabel('Number of participants')
ax.set_title('Participant Overlap Across Tables')
for i, v in enumerate(overlap_data.values()):
    ax.text(v + 1, i, str(v), va='center', fontsize=9)
plt.tight_layout()
plt.show()
```

## 10. Longitudinal Subjects (Multi-Season Participants) <a id='10'></a>

`subject` and `participant_id` are different things. A single `subject` (person) can appear as multiple `participant_id` rows — once per flu season they were enrolled in. This is a key feature of the longitudinal design, but it means observations are **not independent** across seasons for the same subject.

```python
# How many studies does each subject appear in?
subject_study_counts = (
    train_participants
    .groupby('subject')['investigation_id']
    .nunique()
    .sort_values(ascending=False)
)

print(f'Total unique subjects: {len(subject_study_counts):,}')
print(f'\nDistribution of how many studies each subject appears in:')
print(subject_study_counts.value_counts().sort_index().rename('n_subjects').to_string())
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
vc = subject_study_counts.value_counts().sort_index()
ax.bar(vc.index.astype(str), vc.values, color='mediumseagreen', edgecolor='white')
ax.set_xlabel('Number of flu seasons enrolled in')
ax.set_ylabel('Number of subjects')
ax.set_title('Subject Longitudinal Participation')
for i, (x, y) in enumerate(zip(vc.index, vc.values)):
    ax.text(i, y + 2, str(y), ha='center', fontsize=9)
plt.tight_layout()
plt.show()
```

## 12. Flow Cytometry — Unit-Stratified Distributions <a id='12'></a>

The `value` column in `train_flow` mixes different units (`cells/uL`, `percentage`, `percentile`, `ul`, `cells`). Plotting them together would be meaningless — we split by unit. For visualization, values are clipped at the 99th percentile within each unit to prevent extreme outliers from collapsing the scale.


```python
units = train_flow['unit'].dropna().unique()
fig, axes = plt.subplots(1, len(units), figsize=(18, 4))

for ax, unit in zip(axes, units):
    group = train_flow[train_flow['unit'] == unit]['value']
    clip_val = group.quantile(0.99)
    ax.hist(group.clip(upper=clip_val), bins=60, color='steelblue', edgecolor='white', linewidth=0.3)
    ax.set_title(f'unit = {unit}\n(n={len(group):,})', fontsize=9)
    ax.set_xlabel('value (clipped at 99th pct)')
    ax.set_ylabel('Count')

plt.suptitle('train_flow: Value Distributions by Unit', fontsize=12, y=1.02)
plt.tight_layout()
plt.show()
```

## 13. Gene Expression — Distribution by Platform <a id='13'></a>

The `train_array` table has 4 microarray platforms. Even though values are already log2-normalized, the distributions may differ across platforms — a sign of batch effects that would need correction (e.g., ComBat) before combining them for modeling.

```python
platforms = train_array['platform_id'].unique()
fig, axes = plt.subplots(1, len(platforms), figsize=(16, 4))

for ax, platform in zip(axes, platforms):
    group = train_array[train_array['platform_id'] == platform]['value'].dropna()
    ax.hist(group, bins=80, color='darkorange', edgecolor='white', linewidth=0.2)
    ax.set_title(f'{platform}\n(n={len(group):,})', fontsize=9)
    ax.set_xlabel('log2 expression value')
    ax.set_ylabel('Count')

plt.suptitle('train_array: Expression Value Distributions by Platform', fontsize=12, y=1.02)
plt.tight_layout()
plt.show()
```

## 13a. Gene Expression — Deep-Dive <a id='13a'></a>

Going beyond per-platform histograms: we look at how expression changes over time,
which genes are most variable (candidates for feature selection), and whether
platform differences persist after accounting for timepoint.


```python
# ── Mean expression trajectory per participant across timepoints ──────────────
# Each line = one participant's average log2 expression at each timepoint.
# Useful for spotting global shifts in expression (e.g., immune activation at Day 7).

participant_tp_mean = (
    train_array
    .groupby(['participant_id', 'timepoint'])['value']
    .mean()
    .reset_index()
)

fig, ax = plt.subplots(figsize=(10, 5))
for pid, grp in participant_tp_mean.groupby('participant_id'):
    ax.plot(grp['timepoint'], grp['value'], alpha=0.15, color='steelblue', linewidth=0.8)

# Overlay the global mean trajectory
global_mean = participant_tp_mean.groupby('timepoint')['value'].mean()
ax.plot(global_mean.index, global_mean.values, color='crimson', linewidth=2.5, label='Global mean')

ax.set_xlabel('Timepoint (days post-vaccination)')
ax.set_ylabel('Mean log2 expression')
ax.set_title('train_array: Per-Participant Mean Expression Trajectory')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# ── Top 30 most variable genes (by variance across all measurements) ─────────
# High-variance genes carry the most discriminative signal and are strong
# candidates for feature selection before modeling.

gene_var = (
    train_array
    .groupby('gene_symbol')['value']
    .var()
    .sort_values(ascending=False)
)

top_genes = gene_var.head(30)

fig, ax = plt.subplots(figsize=(10, 6))
ax.barh(top_genes.index[::-1], top_genes.values[::-1], color='darkorange', edgecolor='white')
ax.set_xlabel('Variance (log2 expression)')
ax.set_title('Top 30 Most Variable Genes (train_array)')
plt.tight_layout()
plt.show()

print(f'Total genes: {len(gene_var):,}')
print(f'Variance range: {gene_var.min():.4f} — {gene_var.max():.4f}')
print(f'Genes with variance > 1.0: {(gene_var > 1.0).sum():,}')
```

```python
# ── Day 0 → Day 7 expression change (per gene) ────────────────────────────────
# Since values are log2-normalized, the difference (Day7 - Day0) is the log2
# fold-change. A shift away from 0 means the gene was up- or down-regulated
# after vaccination.

day0 = train_array[train_array['timepoint'] == 0].groupby('gene_symbol')['value'].mean()
day7 = train_array[train_array['timepoint'] == 7].groupby('gene_symbol')['value'].mean()
shared_genes = day0.index.intersection(day7.index)
log2fc = (day7.loc[shared_genes] - day0.loc[shared_genes]).sort_values()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: histogram of all fold-changes
ax = axes[0]
ax.hist(log2fc, bins=80, color='steelblue', edgecolor='white', linewidth=0.3)
ax.axvline(0, color='red', linestyle='--', linewidth=1, label='No change')
ax.set_xlabel('log2 fold-change (Day 7 − Day 0)')
ax.set_ylabel('Number of genes')
ax.set_title('Distribution of Day 0 → Day 7 Expression Change')
ax.legend()

# Right: top 15 up-regulated and top 15 down-regulated genes
ax = axes[1]
top_up = log2fc.tail(15)
top_down = log2fc.head(15)
top_both = pd.concat([top_down, top_up])
colors = ['#d62728' if v < 0 else '#2ca02c' for v in top_both.values]
ax.barh(top_both.index, top_both.values, color=colors, edgecolor='white')
ax.axvline(0, color='black', linewidth=0.5)
ax.set_xlabel('log2 fold-change (Day 7 − Day 0)')
ax.set_title('Top 15 Up- & Down-Regulated Genes')

plt.tight_layout()
plt.show()

print(f'Genes with |log2FC| > 0.5: {(log2fc.abs() > 0.5).sum():,} / {len(log2fc):,}')
print(f'Genes with |log2FC| > 1.0: {(log2fc.abs() > 1.0).sum():,} / {len(log2fc):,}')
```

```python
# ── Platform × Timepoint interaction ──────────────────────────────────────────
# Do different platforms show different temporal patterns?
# If the lines aren't parallel, platform and timepoint interact — meaning batch
# correction needs to account for timepoint, not just platform.

platform_tp_mean = (
    train_array
    .groupby(['platform_id', 'timepoint'])['value']
    .mean()
    .reset_index()
)

fig, ax = plt.subplots(figsize=(10, 5))
for platform, grp in platform_tp_mean.groupby('platform_id'):
    ax.plot(grp['timepoint'], grp['value'], marker='o', label=platform, linewidth=1.5)

ax.set_xlabel('Timepoint (days post-vaccination)')
ax.set_ylabel('Mean log2 expression')
ax.set_title('Mean Expression by Platform × Timepoint')
ax.legend(title='Platform')
plt.tight_layout()
plt.show()
```

## 12a. Flow Cytometry — Deep-Dive <a id='12a'></a>

Beyond the unit-stratified histograms: we examine how key cell populations
change over time after vaccination, which populations correlate with each other,
and how much inter-participant variability exists.


```python
# ── Cell population trajectories over time (cells/uL only) ─────────────────
# Restrict to the most common unit (cells/uL) and the top named populations
# to keep the plots interpretable. Each line = one participant.

flow_cells = train_flow[
    (train_flow['unit'] == 'cells/uL') &
    (train_flow['name'].notna()) &
    (train_flow['timepoint'].between(-7, 90))
].copy()

# Pick the 6 most common named populations in this subset
top_pops = flow_cells['name'].value_counts().head(6).index.tolist()

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

for ax, pop in zip(axes, top_pops):
    subset = flow_cells[flow_cells['name'] == pop]
    for pid, grp in subset.groupby('participant_id'):
        grp_sorted = grp.sort_values('timepoint')
        ax.plot(grp_sorted['timepoint'], grp_sorted['value'],
                alpha=0.15, color='steelblue', linewidth=0.7)

    # Overlay median trajectory
    med = subset.groupby('timepoint')['value'].median()
    ax.plot(med.index, med.values, color='crimson', linewidth=2, label='Median')

    ax.set_title(pop, fontsize=9)
    ax.set_xlabel('Day')
    ax.set_ylabel('cells/uL')
    ax.legend(fontsize=7)

plt.suptitle('train_flow: Cell Population Trajectories (cells/uL)', fontsize=13, y=1.01)
plt.tight_layout()
plt.show()
```

```python
# ── Violin plots: cell population distributions at Day 0, 7, 28 ─────────────
# Shows distribution shape + spread at each core timepoint.
# Differences in shape across timepoints indicate immune response dynamics.

core_days = [0.0, 7.0, 28.0]
flow_core = flow_cells[
    (flow_cells['timepoint'].isin(core_days)) &
    (flow_cells['name'].isin(top_pops))
].copy()
flow_core['day_label'] = 'Day ' + flow_core['timepoint'].astype(int).astype(str)

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
axes = axes.flatten()

for ax, pop in zip(axes, top_pops):
    subset = flow_core[flow_core['name'] == pop]
    if subset.empty:
        ax.set_visible(False)
        continue
    # Clip at 99th percentile within this population for visibility
    clip_val = subset['value'].quantile(0.99)
    subset = subset.copy()
    subset['value_clipped'] = subset['value'].clip(upper=clip_val)

    day_order = sorted(subset['day_label'].unique())
    sns.violinplot(
        data=subset, x='day_label', y='value_clipped',
        order=day_order, ax=ax, inner='quartile',
        palette='Set2', cut=0
    )
    ax.set_title(pop, fontsize=9)
    ax.set_xlabel('')
    ax.set_ylabel('cells/uL (clipped 99th pct)')

plt.suptitle('train_flow: Cell Populations at Core Timepoints', fontsize=13, y=1.01)
plt.tight_layout()
plt.show()
```

```python
# ── Correlation between cell populations ─────────────────────────────────────
# Pivot to wide format (participant × timepoint → population values) and compute
# pairwise Pearson correlation. High correlations may indicate redundant features.

# Use Day 0, cells/uL, named populations with enough data
flow_d0 = flow_cells[
    (flow_cells['timepoint'] == 0.0) &
    (flow_cells['name'].notna())
].copy()

pop_counts = flow_d0['name'].value_counts()
valid_pops = pop_counts[pop_counts >= 20].index  # need enough observations
flow_d0 = flow_d0[flow_d0['name'].isin(valid_pops)]

# Pivot: rows = participants, columns = population names
flow_wide = flow_d0.pivot_table(
    index='participant_id', columns='name', values='value', aggfunc='mean'
)

# Drop columns with too many NaNs
flow_wide = flow_wide.dropna(axis=1, thresh=int(len(flow_wide) * 0.5))

if flow_wide.shape[1] >= 3:
    corr = flow_wide.corr()
    fig, ax = plt.subplots(figsize=(12, 10))
    sns.heatmap(
        corr, ax=ax, cmap='RdBu_r', center=0, vmin=-1, vmax=1,
        annot=True, fmt='.2f', annot_kws={'size': 7},
        linewidths=0.5, square=True
    )
    ax.set_title('Cell Population Correlation (Day 0, cells/uL)', fontsize=12)
    ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha='right', fontsize=8)
    ax.set_yticklabels(ax.get_yticklabels(), fontsize=8)
    plt.tight_layout()
    plt.show()
else:
    print('Not enough cell populations with sufficient data for correlation analysis.')
```

```python
# ── Inter-participant variability by cell population (Day 0) ──────────────────
# Coefficient of variation (CV = std/mean) shows which populations are most
# variable relative to their scale. High-CV populations may be more informative
# as features but also noisier.

if flow_wide.shape[1] >= 3:
    cv = flow_wide.std() / flow_wide.mean()
    cv = cv.dropna().sort_values(ascending=True)

    fig, ax = plt.subplots(figsize=(8, max(4, len(cv) * 0.35)))
    colors = ['#d62728' if v > 1.0 else 'steelblue' for v in cv.values]
    ax.barh(cv.index, cv.values, color=colors, edgecolor='white')
    ax.axvline(1.0, color='grey', linestyle='--', linewidth=0.8, label='CV = 1.0')
    ax.set_xlabel('Coefficient of Variation')
    ax.set_title('Inter-Participant Variability by Cell Population (Day 0, cells/uL)')
    ax.legend()
    plt.tight_layout()
    plt.show()
```

```python
# How many participants have data at each 'core' timepoint?
# The effective sample size at each timepoint is what matters for modeling.

core_tps_array = [-14, 0, 1, 3, 7, 28]
core_tps_flow = [-7.0, 0.0, 1.0, 7.0, 28.0, 70.0, 90.0]  # float to match dtype

print('Participants with data at each core timepoint (train_array):')
for tp in core_tps_array:
    n = train_array[train_array['timepoint'] == tp]['participant_id'].nunique()
    print(f'  Day {tp:>4}: {n:>4} participants')

print('\nParticipants with data at each core timepoint (train_flow):')
for tp in core_tps_flow:
    n = train_flow[train_flow['timepoint'] == tp]['participant_id'].nunique()
    print(f'  Day {tp:>5.0f}: {n:>4} participants')
```

```python
# Which timepoints are shared between train_array and train_flow?
# These are the timepoints where multi-modal analysis (gene expression + cell populations)
# is possible — important for feature engineering.
#
# Note: train_array timepoints are int64, train_flow are float64.
# We normalize both to float for safe comparison.

array_tps = set(float(t) for t in train_array['timepoint'].unique())
flow_tps = set(float(t) for t in train_flow['timepoint'].unique())
shared_tps = sorted(array_tps & flow_tps)
only_array = sorted(array_tps - flow_tps)
only_flow = sorted(flow_tps - array_tps)

print(f'Timepoints in BOTH tables (multi-modal possible): {[int(t) if t == int(t) else t for t in shared_tps]}')
print(f'Timepoints only in train_array:  {[int(t) if t == int(t) else t for t in only_array]}')
print(f'Timepoints only in train_flow:   {[int(t) if t == int(t) else t for t in only_flow]}')
```

```python
# Heatmap: participant × timepoint coverage for train_array
# Each cell = number of genes measured for that participant at that timepoint.
# Gaps (white) = missing data — important for imputation decisions.

# Pivot: rows = participants, cols = timepoints, values = gene measurement count
array_coverage = (
    train_array
    .groupby(['participant_id', 'timepoint'])['gene_symbol']
    .count()
    .unstack(fill_value=0)
)

# Binarize for readability: 1 = has data, 0 = missing
array_coverage_bin = (array_coverage > 0).astype(int)

fig, ax = plt.subplots(figsize=(14, max(6, len(array_coverage_bin) * 0.18)))
sns.heatmap(
    array_coverage_bin,
    ax=ax,
    cmap='Blues',
    cbar=False,
    linewidths=0.2,
    linecolor='lightgrey'
)
ax.set_title('train_array: Participant × Timepoint Coverage\n(blue = has data, white = missing)', fontsize=11)
ax.set_xlabel('Timepoint (days)')
ax.set_ylabel('Participant')
ax.set_yticklabels(ax.get_yticklabels(), fontsize=5)
plt.tight_layout()
plt.show()

# Summary: what fraction of participants have complete core timepoints?
core = [0, 7, 28]
has_core = array_coverage_bin[core].all(axis=1)
print(f'\nParticipants with data at ALL of days {core}: {has_core.sum()} / {len(has_core)}')
```

```python
# Same coverage heatmap for train_flow

flow_coverage = (
    train_flow
    .groupby(['participant_id', 'timepoint'])['flow_id']
    .count()
    .unstack(fill_value=0)
)
flow_coverage_bin = (flow_coverage > 0).astype(int)

fig, ax = plt.subplots(figsize=(14, max(6, len(flow_coverage_bin) * 0.18)))
sns.heatmap(
    flow_coverage_bin,
    ax=ax,
    cmap='Oranges',
    cbar=False,
    linewidths=0.2,
    linecolor='lightgrey'
)
ax.set_title('train_flow: Participant × Timepoint Coverage\n(orange = has data, white = missing)', fontsize=11)
ax.set_xlabel('Timepoint (days)')
ax.set_ylabel('Participant')
ax.set_yticklabels(ax.get_yticklabels(), fontsize=5)
plt.tight_layout()
plt.show()

# Float literals to match the float64 column index from unstack
core_flow = [0.0, 7.0, 28.0]
available_core = [c for c in core_flow if c in flow_coverage_bin.columns]
has_core_flow = flow_coverage_bin[available_core].all(axis=1)
print(
    f'\nParticipants with data at ALL of days {[int(c) for c in available_core]}: {has_core_flow.sum()} / {len(has_core_flow)}')
```

## 16. EDA Summary & Key Findings <a id='16'></a>

### What we found

| Finding | Detail | Implication |
|---|---|---|
| **Very limited multi-modal overlap** | Only **45 participants** appear in all 3 tables | Most modeling will need to treat array and flow as separate datasets unless you restrict to this small subset |
| **train_array is dominated by days 0 and 7** | Day 0: 208 participants, Day 7: 200 — but Day 28: only 74 | Many participants are missing late timepoints; imputation or analysis restricted to days 0–7 may be needed |
| **train_flow has better longitudinal coverage** | 176 of 264 participants have days 0, 7, and 28 | Flow data is more complete for longitudinal modeling |
| **train_participants is metadata-only** | 3,960 rows, but only 208 (array) and 264 (flow) have assay data | Useful for demographic stratification but cannot stand alone for prediction |
| **4 microarray platforms in train_array** | GPL10558 dominates (~8.6M rows); GPL6947 is tiny (~6k rows) | Batch correction (e.g., ComBat) is likely needed before cross-platform modeling |
| **Mixed units in train_flow** | 5 unit types: `cells/uL`, `percentage`, `percentile`, `ul`, `cells` | Always filter by `unit` before aggregating; never mix units in a model feature |
| **train_participants skews female, mostly Unknown race** | 59% female; 54% race Unknown | Class imbalance in demographics; Unknown race limits subgroup analyses |
| **Longitudinal subjects** | 577 subjects appear in 2+ flu seasons (up to 8) | Must use subject-level grouping for cross-validation splits — **do not split by participant_id** |

### Recommended next steps
- **Batch correction**: Apply ComBat or similar to `train_array` across the 4 platforms before feature extraction
- **Unit-stratified analysis**: Always keep `unit` as a filter when working with `train_flow`
- **Cross-validation strategy**: Group by `subject` (not `participant_id`) when doing train/val splits to prevent data leakage across flu seasons
- **Dimensionality reduction**: With 39,665 genes × 208 participants, consider filtering to high-variance genes before pivoting to wide format
